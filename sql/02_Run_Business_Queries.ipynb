## Query 4 – Average Salary by Department

### Objective

Identify which departments pay the highest average monthly salary.

```python
query = """
SELECT
    Department,
    ROUND(AVG(MonthlyIncome),2) AS AverageSalary
FROM employees
GROUP BY Department
ORDER BY AverageSalary DESC;
"""

pd.read_sql(query, conn)
```

## Query 5 – Attrition by Department

### Objective

Count employee attrition across departments.

```python
query = """
SELECT
    Department,
    Attrition,
    COUNT(*) AS Employees
FROM employees
GROUP BY Department, Attrition
ORDER BY Department;
"""

pd.read_sql(query, conn)
```

## Query 6 – Average Age by Department

### Objective

Find the average employee age in each department.

```python
query = """
SELECT
    Department,
    ROUND(AVG(Age),1) AS AverageAge
FROM employees
GROUP BY Department
ORDER BY AverageAge DESC;
"""

pd.read_sql(query, conn)
```

## Query 21 – Large Departments

### Objective

Find departments having more than 300 employees.

```python
query = """
SELECT
    Department,
    COUNT(*) AS Employees
FROM employees
GROUP BY Department
HAVING COUNT(*) > 300;
"""

pd.read_sql(query, conn)
```

## Query 22 – Employees Above Average Salary

### Objective

Find employees earning above the company average salary.

```python
query = """
SELECT
    EmployeeNumber,
    Department,
    MonthlyIncome
FROM employees
WHERE MonthlyIncome >
(
    SELECT AVG(MonthlyIncome)
    FROM employees
)
ORDER BY MonthlyIncome DESC;
"""

pd.read_sql(query, conn)
```

## Query 24 – Department Salary Ranking

### Objective

Rank departments by average salary.

## Query 25 – Salary Ranking

### Objective

Rank employees based on monthly income.

```python
query = """
SELECT
    EmployeeNumber,
    MonthlyIncome,
    RANK() OVER(
        ORDER BY MonthlyIncome DESC
    ) AS SalaryRank
FROM employees;
"""

pd.read_sql(query, conn)
```

## Query 26 – Top Paid Employees

### Objective

Display the highest paid employees.

```python
query = """
SELECT
    EmployeeNumber,
    Department,
    MonthlyIncome
FROM employees
ORDER BY MonthlyIncome DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Query 27 – Salary Rank Within Department

### Objective

Rank employees by salary inside each department.

```python
query = """
SELECT
    EmployeeNumber,
    Department,
    MonthlyIncome,
    RANK() OVER(
        PARTITION BY Department
        ORDER BY MonthlyIncome DESC
    ) AS DepartmentRank
FROM employees;
"""

pd.read_sql(query, conn)
```

## Query 28 – Running Average Salary

### Objective

Calculate cumulative average salary.

```python
query = """
SELECT
    EmployeeNumber,
    MonthlyIncome,
    AVG(MonthlyIncome) OVER(
        ORDER BY MonthlyIncome
    ) AS RunningAverage
FROM employees;
"""

pd.read_sql(query, conn)
```

## Query 29 – Dense Salary Ranking

### Objective

Assign dense ranks based on salary.

```python
query = """
SELECT
    EmployeeNumber,
    MonthlyIncome,
    DENSE_RANK() OVER(
        ORDER BY MonthlyIncome DESC
    ) AS DenseSalaryRank
FROM employees;
"""

pd.read_sql(query, conn)
```

## Query 30 – Employee Numbering by Department

### Objective

Assign row numbers within each department.

```python
query = """
SELECT
    EmployeeNumber,
    Department,
    ROW_NUMBER() OVER(
        PARTITION BY Department
        ORDER BY MonthlyIncome DESC
    ) AS RowNum
FROM employees;
"""

pd.read_sql(query, conn)
```
