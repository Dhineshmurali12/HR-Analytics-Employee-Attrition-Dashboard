# HR Analytics Employee Attrition Dashboard

# Day 4 – Business-Oriented Analysis

## Objective

The objective of this notebook is to analyze employee attrition from a business perspective by identifying the factors that contribute to employee turnover.

### Business Questions

- Which department has the highest attrition?
- Which job roles experience the highest turnover?
- Does overtime influence employee attrition?
- Does gender impact attrition?
- Does marital status affect retention?
- Does business travel contribute to attrition?
- Which education fields experience higher attrition?
- Which age groups are more likely to leave?
- Does salary influence employee retention?

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

plt.style.use("default")
```

```python
df = pd.read_csv("../data/hr_cleaned.csv")

df.head()
```

## 1. Attrition by Department

This analysis identifies departments with higher employee turnover.

```python
department_attrition = (
    pd.crosstab(
        df["Department"],
        df["Attrition"],
        normalize="index"
    ) * 100
).round(2)

department_attrition
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition",
    palette="Set2"
)

plt.title("Employee Attrition by Department")

plt.show()
```

### Insight

- Compare attrition across departments.
- Departments with higher attrition may require targeted retention strategies.

```python
bins = [18,25,35,45,55,65]

labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65"
]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
```

```python
age_attrition = (
    pd.crosstab(
        df["AgeGroup"],
        df["Attrition"],
        normalize="index"
    ) * 100
).round(2)

age_attrition
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="AgeGroup",
    hue="Attrition",
    palette="Set2"
)

plt.title("Attrition by Age Group")

plt.show()
```

```python
df["IncomeBand"] = pd.qcut(
    df["MonthlyIncome"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
income_attrition = (
    pd.crosstab(
        df["IncomeBand"],
        df["Attrition"],
        normalize="index"
    ) * 100
).round(2)

income_attrition
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="IncomeBand",
    hue="Attrition",
    palette="Set2"
)

plt.title("Attrition by Income Band")

plt.show()
```
