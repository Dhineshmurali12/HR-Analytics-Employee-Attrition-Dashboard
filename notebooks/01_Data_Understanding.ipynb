## Attrition Distribution

This visualization shows the proportion of employees who stayed versus those who left the company.

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Attrition",
    palette="Set2"
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

## Age Distribution

Understanding the age distribution of employees helps identify the workforce composition and whether attrition is concentrated among specific age groups.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True,
    color="steelblue"
)

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

### Insight

- Most employees are between **30–40 years old**.
- The workforce is primarily mid-career professionals.
- Younger and older employees represent a smaller proportion of the workforce.

## Monthly Income Distribution

Monthly income analysis helps understand salary distribution and identify potential income-related attrition patterns.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="MonthlyIncome",
    bins=30,
    kde=True,
    color="green"
)

plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")

plt.show()
```

### Insight

- Salary distribution is right-skewed.
- Most employees earn lower to mid-range salaries.
- A small number of employees receive significantly higher salaries.

## Distance From Home Distribution

This analysis helps determine whether employees generally live close to or far from the workplace.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="DistanceFromHome",
    bins=20,
    kde=True,
    color="orange"
)

plt.title("Distance From Home Distribution")
plt.xlabel("Distance From Home")
plt.ylabel("Employees")

plt.show()
```

### Insight

- Most employees live relatively close to their workplace.
- A smaller group commutes longer distances.
- Distance may contribute to employee attrition when combined with other factors.

## Job Satisfaction Distribution

Job satisfaction is an important factor influencing employee retention.

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue="JobSatisfaction",
    palette="Blues",
    legend=False
)

plt.title("Job Satisfaction Levels")
plt.xlabel("Job Satisfaction")
plt.ylabel("Employees")

plt.show()
```

# Outlier Detection

Boxplots help identify extreme values (outliers) in numerical features such as Monthly Income, Age, Daily Rate, and Distance From Home.

```python
plt.figure(figsize=(8,4))

sns.boxplot(
    x=df["MonthlyIncome"],
    color="lightgreen"
)

plt.title("Monthly Income Boxplot")

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.boxplot(
    x=df["Age"],
    color="skyblue"
)

plt.title("Age Boxplot")

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.boxplot(
    x=df["DistanceFromHome"],
    color="orange"
)

plt.title("Distance From Home Boxplot")

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.boxplot(
    x=df["DailyRate"],
    color="salmon"
)

plt.title("Daily Rate Boxplot")

plt.show()
```

### Insights

- Monthly Income contains several high-value outliers representing senior employees or executives.
- Age distribution contains very few extreme values.
- Daily Rate appears relatively evenly distributed.
- Distance From Home has employees with longer commute distances.

```python
# Remove constant columns
corr_df = df[numerical_cols].drop(
    columns=["EmployeeCount", "StandardHours"],
    errors="ignore"
)

plt.figure(figsize=(14,10))

sns.heatmap(
    corr_df.corr(),
    cmap="coolwarm",
    annot=False,
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

### Insights

- Most HR variables show weak correlations.
- Monthly Income strongly correlates with Job Level and Total Working Years.
- Weak correlations indicate that employee attrition depends on multiple factors rather than a single variable.
