## Attrition Distribution

This visualization shows the proportion of employees who stayed versus those who left the company.

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Attrition",
    palette="Set2"
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

## Age Distribution

Understanding the age distribution of employees helps identify the workforce composition and whether attrition is concentrated among specific age groups.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True,
    color="steelblue"
)

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

### Insight

- Most employees are between **30–40 years old**.
- The workforce is primarily mid-career professionals.
- Younger and older employees represent a smaller proportion of the workforce.

## Monthly Income Distribution

Monthly income analysis helps understand salary distribution and identify potential income-related attrition patterns.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="MonthlyIncome",
    bins=30,
    kde=True,
    color="green"
)

plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")

plt.show()
```

### Insight

- Salary distribution is right-skewed.
- Most employees earn lower to mid-range salaries.
- A small number of employees receive significantly higher salaries.

## Distance From Home Distribution

# Business Analysis

## Attrition by Department

This analysis identifies which departments experience the highest employee attrition.

```python
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

department_attrition
```

```python
department_attrition_percent = pd.crosstab(
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100

department_attrition_percent.round(2)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition",
    palette="Set2"
)

plt.title("Employee Attrition by Department")
plt.xticks(rotation=15)

plt.show()
```

### Insight

- Compare the number of employees leaving each department.
- Identify departments requiring better retention strategies.
- Departments with higher attrition may require management attention.

## Attrition by Job Role

Job roles often have different turnover rates due to workload, salary, and career growth opportunities.

```python
jobrole_attrition = pd.crosstab(
    df["JobRole"],
    df["Attrition"],
    normalize="index"
) * 100

jobrole_attrition.round(2)
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    y="JobRole",
    hue="Attrition",
    palette="Set2"
)

plt.title("Attrition by Job Role")

plt.show()
```

### Insight

- These positions may require salary reviews, career progression plans, or workload balancing.

```python
bins = [18, 25, 35, 45, 55, 65]

labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65"
]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
```

```python
agegroup_attrition = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

agegroup_attrition.round(2)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="AgeGroup",
    hue="Attrition",
    palette="Set2"
)

plt.title("Attrition by Age Group")

plt.show()
```

```python
df["IncomeBand"] = pd.qcut(
    df["MonthlyIncome"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
income_attrition = pd.crosstab(
    df["IncomeBand"],
    df["Attrition"],
    normalize="index"
) * 100

income_attrition.round(2)
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="IncomeBand",
    hue="Attrition",
    palette="Set2"
)

plt.title("Attrition by Income Band")

plt.show()
```
